Create a simple Python program called **Purchase Advisor** that helps a user decide whether they should buy an item based on their monthly budget.

The program should:

1. Ask the user to enter their monthly budget.
2. Ask how much money they have already spent this month.
3. Ask for the price of the item they want to buy.
4. Calculate the remaining budget by subtracting the amount already spent from the monthly budget.
5. Calculate what percentage of the remaining budget the item would cost.
6. Use `if`, `elif`, and `else` statements to give one of these decisions:

   * If the item costs more than the remaining budget, print **"Cannot afford this purchase."**
   * If the item costs 20% or less of the remaining budget, print **"Safe to buy."**
   * If the item costs more than 20% but no more than 50% of the remaining budget, print **"Think carefully before buying."**
   * If the item costs more than 50% of the remaining budget but does not exceed it, print **"Not recommended."**
7. Display the remaining budget before the purchase, the item price, and the remaining budget after the decision. If the item costs more than the remaining budget, do not subtract the item price and keep the remaining budget unchanged.
8. Use `float()` for money inputs and display monetary amounts to two decimal places.
9. Keep the program beginner-friendly. Use only variables, `input()`, `print()`, arithmetic operations, and `if`/`elif`/`else`. Do not use functions, loops, lists, dictionaries, classes, or external libraries.
10. Produce the complete runnable Python code without asking any follow-up questions.


````python
# Paste the code your prompt produced here, then run it to show it works.
print("Purchase Advisor")

monthly_budget = float(input("Enter your monthly budget: $"))
amount_spent = float(input("Enter how much you have already spent this month: $"))
item_price = float(input("Enter the price of the item: $"))

remaining_budget = monthly_budget - amount_spent

print(f"\nRemaining budget before purchase: ${remaining_budget:.2f}")
print(f"Item price: ${item_price:.2f}")

if item_price > remaining_budget:
    print("Cannot afford this purchase.")
    remaining_after_purchase = remaining_budget

else:
    percentage = (item_price / remaining_budget) * 100

    if percentage <= 20:
        print("Safe to buy.")
    elif percentage <= 50:
        print("Think carefully before buying.")
    else:
        print("Not recommended.")

    remaining_after_purchase = remaining_budget - item_price

print(f"Remaining budget after decision: ${remaining_after_purchase:.2f}")
````
